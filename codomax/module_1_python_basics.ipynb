```python
# 1. VARIABLES & DATA TYPES
# Storing details for a single delivery truck trip
truck_id = "TN-01-AB-1234"       # String
distance_km = 245.5              # Float
fuel_efficiency_kmpl = 6.5       # Float (kilometers per liter)
fuel_price_per_liter = 94.50     # Float (in INR)
is_express_delivery = True       # Boolean

# Calculating total fuel needed and total trip fuel cost
fuel_needed = distance_km / fuel_efficiency_kmpl
total_fuel_cost = fuel_needed * fuel_price_per_liter

# Using an f-string to print a clean summary
print(f"Truck {truck_id} traveled {distance_km} km.")
print(f"Estimated Fuel Needed: {fuel_needed:.2f} liters")
print(f"Total Fuel Cost: ₹{total_fuel_cost:.2f}")
```

```python
# 2. LOOPS & CONDITIONALS
# Delivery times (in minutes) for 7 shipments today
delivery_times = [28, 45, 32, 60, 25, 52, 30]
sla_limit = 40  # Service Level Agreement: deliveries over 40 mins are "Delayed"

total_time = 0
delayed_count = 0

for time in delivery_times:
    total_time += time
    if time > sla_limit:
        delayed_count += 1
        print(f"Delivery took {time} mins -> Status: DELAYED")
    else:
        print(f"Delivery took {time} mins -> Status: On-Time")

average_time = total_time / len(delivery_times)
print("-" * 40)
print(f"Average Delivery Time: {average_time:.1f} minutes")
print(f"Total Delayed Shipments: {delayed_count} out of {len(delivery_times)}")
```

```python
# 3. FUNCTIONS
def calculate_shipping_cost(weight_kg, is_express=False):
    """
    Calculates shipping cost based on package weight and delivery speed.
    Base rate: ₹50 + ₹20 per kg. Express adds a 50% surcharge.
    """
    base_rate = 50
    per_kg_rate = 20
    
    cost = base_rate + (weight_kg * per_kg_rate)
    
    if is_express:
        cost = cost * 1.5  # Add 50% express fee
        
    return cost

# Testing the function with different shipments
pkg_1 = calculate_shipping_cost(weight_kg=3.5, is_express=False)
pkg_2 = calculate_shipping_cost(weight_kg=10.0, is_express=True)

print(f"Standard Package (3.5 kg): ₹{pkg_1:.2f}")
print(f"Express Package (10.0 kg): ₹{pkg_2:.2f}")
```
