```python
# 1. LISTS & LIST COMPREHENSIONS
# List of package weights (in kg) loaded onto a delivery van
package_weights = [2.5, 14.0, 5.2, 22.8, 8.0, 18.5, 1.2]

# Basic list operations
package_weights.append(6.4)   # Add a late arrival package
package_weights.sort()        # Sort from lightest to heaviest

print(f"Sorted Weights (kg): {package_weights}")
print(f"Lightest Package: {package_weights[0]} kg | Heaviest Package: {package_weights[-1]} kg")

# List comprehension: Filter out heavy packages (> 10 kg) that require two-person handling
heavy_packages = [w for w in package_weights if w > 10.0]
print(f"Heavy Packages (>10 kg): {heavy_packages}")
```

```python
# 2. DICTIONARIES (STRUCTURED RECORDS)
shipments = [
    {"tracking_id": "TRK101", "destination": "Chennai", "weight_kg": 4.5, "status": "Delivered"},
    {"tracking_id": "TRK102", "destination": "Bengaluru", "weight_kg": 12.0, "status": "In Transit"},
    {"tracking_id": "TRK103", "destination": "Hyderabad", "weight_kg": 2.1, "status": "Delayed"},
    {"tracking_id": "TRK104", "destination": "Mumbai", "weight_kg": 18.5, "status": "In Transit"},
    {"tracking_id": "TRK105", "destination": "Chennai", "weight_kg": 6.0, "status": "Delayed"}
]

# Loop through the dictionary records to flag delayed shipments
print("⚠️ DELAYED SHIPMENTS ALERT:")
for item in shipments:
    if item["status"] == "Delayed":
        print(f" - ID: {item['tracking_id']} | Destination: {item['destination']} ({item['weight_kg']} kg)")

# Calculate total cargo weight across all shipments
total_cargo = sum(item["weight_kg"] for item in shipments)
print(f"\nTotal Fleet Cargo Weight: {total_cargo} kg")
```

```python
# 3. FILE HANDLING: WRITING TO CSV
import csv

filename = "shipments_data.csv"
headers = ["tracking_id", "destination", "weight_kg", "status"]

# 'w' mode opens the file for writing; newline='' prevents blank lines on Windows
with open(filename, mode="w", newline="") as file:
    writer = csv.DictWriter(file, fieldnames=headers)
    writer.writeheader()       # Writes the column names
    writer.writerows(shipments) # Writes all dictionary rows

print(f"✅ Successfully saved {len(shipments)} records to '{filename}'!")
```

```python
# 4. FILE HANDLING: READING CSV & WRITING SUMMARY REPORT
read_records = []

with open("shipments_data.csv", mode="r") as file:
    reader = csv.DictReader(file)
    for row in reader:
        # Convert weight back from string to float since CSVs store everything as text
        row["weight_kg"] = float(row["weight_kg"])
        read_records.append(row)

# Generate a formatted text summary file
with open("logistics_summary.txt", mode="w") as report:
    report.write("DAILY LOGISTICS OPERATIONS SUMMARY\n")
    report.write("=" * 35 + "\n")
    for rec in read_records:
        line = f"{rec['tracking_id']} -> {rec['destination']} | Status: {rec['status']}\n"
        report.write(line)

print("✅ Generated 'logistics_summary.txt'. Here is its content:\n")
with open("logistics_summary.txt", mode="r") as report:
    print(report.read())
```
